```python
from cirq_sic import *
```

```python
base_dir = "data/sky_ground"
d = 2
n_states = 100
states = [rand_ket(d) for _ in range(n_states)]
wh_implementation = "ak"
specs = {"dataset_id": "negativity",
         "processor_id": "willow_pink",
         "run_type": "clean",
         "qubits": get_wh_qubits(d, wh_implementation),
         "n_shots": 50000,
         "optimizer": "cirq",
         "d": d,
         "fiducial": rand_ket(d), # load_sic_fiducial(d),
         "fiducial_description": "rand_ket", # "numerical_sic",
         "wh_implementation": wh_implementation}
rand_states_task = WHPOVMOnStatesTask(**{**specs, "states": states, "states_description": "rand_ket"})
run_sky_ground_task(rand_states_task, base_dir=base_dir)
characterize_task = CharacterizeWHReferenceDeviceTask(**specs)
run_sky_ground_task(characterize_task, base_dir=base_dir)
```

```python
r = np.array(load_results(rand_states_task, base_dir=base_dir)["processed_data"]["r"])
P = np.array(load_results(characterize_task, base_dir=base_dir)["processed_data"]["P"])
Phi = np.linalg.inv(P)
quasi_probabilities = Phi @ r
avg_negativity = np.sum((abs(quasi_probabilities) - quasi_probabilities)/2)/r.shape[1]
avg_negativity
```
