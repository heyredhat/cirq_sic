```python
from pathlib import Path
import os

from cirq_sic import *
```

```python
def collect_circuit_files(base_dir: str | Path) -> dict[Path, list[str]]:
    """Return {circuits_dir_path: [file names]} for every 'circuits' folder under base_dir."""
    base_path = Path(base_dir).expanduser().resolve()
    circuit_dirs: dict[Path, list[str]] = {}

    for dirpath, dirnames, filenames in os.walk(base_path):
        if Path(dirpath).name == "circuits":
            circuit_dirs[Path(dirpath)] = sorted(filenames)

    return circuit_dirs
```

```python
circuit_files = collect_circuit_files("../experiments/runners/data")
circuits = {}
for k, v in circuit_files.items():
    before, during, after = str(k).partition("data/")
    circuits[after] = {f: cirq.read_json(k / f) for f in v}
```

```python
for task_spec, circs in circuits.items():
    print(task_spec)
    for circ_type, circuit in circs.items():
        print(f"\t{circ_type}:")
        get_gate_counts(circuit[0])
    print("******")
```
